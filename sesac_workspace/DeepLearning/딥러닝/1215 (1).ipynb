````python
messages = [{'role':'system', 'content':'You are a helpful assistant.'},
            {'role':'user','content':'안녕하세요! 저는 시오닝이라고 합니다.'}]
````

````python
# GPT 응답을 한번에 받지 않고 실시간으로 조금식 받는 코드
response = client.chat.completions.create(model='gpt-4o', messages=messages, stream=True)
````

```python
response = client.chat.completions.create(model='gpt-4o', messages=messages, stream=True)
```
-> GPT가 답변을 다 만들때까지 기다리지 말고, 타이핑 치듯이 조각(chunk) 단위로 바로바로 보내달라
- stream=True : GPT가 단어/문장 조각을 생성하는 즉시 하나씩 계속 보내줌(=실시간 출력 가능)
- Stream=True 사용시 response는?  
-> 완성된 결과 객체가 아님 => iterator(이터레이터)  
-> for chunk in response 같은 반복문으로 받아야 함


````python
for chunk in response:
  print(chunk)
````

````python
for chunk in response:
  print(chunk.choices[0].delta.content, end="")
````

end="" 하는 이유!
- print()는 기본적으로 출력 끝에 \n(줄바꿈)이 포함되어있음  
-> ex) print("안", end="\n")

- end="" 사용하면 줄바꿈 없이 이어서 출력됨

````python
# 마지막 None 제외하고 출력하기
for chunk in response:
  content = chunk.choices[0].delta.content
  if content is not None:
    print(content,end="")
````

````python
import json


def get_current_weather(location, unit="Celsius"):
    if "seoul" in location.lower():
        return json.dumps({"location": "Seoul", "temperature": "10", "unit": unit})
    elif "san francisco" in location.lower():
        return json.dumps(
            {"location": "San Francisco", "temperature": "72", "unit": unit}
        )
    elif "paris" in location.lower():
        return json.dumps({"location": "Paris", "temperature": "22", "unit": unit})
    else:
        return json.dumps({"location": location, "temperature": "unknown"})
````
