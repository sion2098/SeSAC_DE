```python
'''

'''

from sklearn.model_selection import train_test_split # 데이터셋 나누기 라이브러리 import
from sklearn.preprocessing import StandardScaler, LabelEncoder #데이터 전처리(스케일링, 인코딩)
from sklearn.neighbors import KNeighborsClassifier # KNN(가장 가까운 이웃 k개 기준으로 클래스 결정_과반수)
from sklearn.metrics import accuracy_score # 정확도 점수
import matplotlib.pyplot as plt # 그래프
import pandas as pd # 판다스

plt.rcParams['font.family'] ='Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False  


df = pd.read_csv('data/iris.csv') #데이터셋 읽어오기

X = df.drop('species', axis = 1)  #df에서 'spexies' 열 제거
y = df['species'] # 'species' 추출


X_train, X_test, y_train, y_test = train_test_split(   # 훈련용/테스트용 데이터 분리
    X, y,  # 데이터
    test_size=0.2,  # 8:2로 데이터 분리
    random_state=42, # 랜덤성 고정
    stratify=y # 클래스 균등하게 선택(써보기)
)


encoder = LabelEncoder() #라벨인코딩
y_train_encoded = encoder.fit_transform(y_train) 
y_test_encoded = encoder.transform(y_test) 

scaler = StandardScaler() 
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)


k_list = range(3, 12, 2) # 3,5,7,9,11
test_accuracies = [] # 리스트 선언

for k in k_list: 
 
    model = KNeighborsClassifier(n_neighbors=k)
    model.fit(X_train_scaled, y_train_encoded)
    
    
    y_test_pred = model.predict(X_test_scaled)
    
    
    test_acc = accuracy_score(y_test_encoded, y_test_pred)
    test_accuracies.append(test_acc) 


best_k_index = test_accuracies.index(max(test_accuracies))
best_k = k_list[best_k_index]

print("K 값별 Test 정확도 :", test_accuracies)
print(f"최적의 K 값: {best_k}")
print(f"최적 K에서의 Test 정확도: {test_accuracies[best_k_index]:.4f}")


best_model = KNeighborsClassifier(n_neighbors=best_k)
best_model.fit(X_train_scaled, y_train_encoded)


y_test_pred_final = best_model.predict(X_test_scaled)
final_test_acc = accuracy_score(y_test_encoded, y_test_pred_final) 

print(f"최종 모델(최적 K={best_k}) Test 정확도: {final_test_acc:.4f}")


plt.figure(figsize=(8, 5))

plt.plot(k_list,test_accuracies , marker='s', label='Test Accuracy')
plt.xticks(k_list)
plt.xlabel('K (Number of Neighbors)')
plt.ylabel('Accuracy')
plt.title('K 값에 따른 KNN 모델 정확도 변화')
plt.legend()
plt.grid(True)

plt.show()
```
