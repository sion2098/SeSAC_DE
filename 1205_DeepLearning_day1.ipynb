```python
# 어파인 변환을 사용한 AND 퍼셉트 구현
def AND(x1, x2):
  result = 0.5*x1 + 0.5*x2 - 0.7  # 어파인 변환
  if result <= 0:
    return 0
  else:
    return 1
```

```python
import numpy as np
data = np.random.random((1000,100))
labels = np.random.randint(2, size=(1000,1))
```

```python
data.shape
```

```python
data
```

```python
labels.shape
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Activation
```

```python
model = Sequential()
model.add(Dense(32, activation='relu', input_dim=100)) # 100인 이유는 이미지 사이즈 10x10으로 100 넣은거임
model.add(Dense(1, activation='sigmoid'))
model.compile(optimizer = 'rmsprop', loss='binary_crossentropy',metrics=['accuracy'])
# metrics는 모델을 잘 돌리고 있는지 중간에 평가하는 것임
```

```python
# 강의자료 p.58 한번 봐바 이해해봐
# 잘못 예측하면 커지고? 뭐지
```

```python
model.fit(data, labels, epochs=10, batch_size=32)
```

```python
#(100,784) -> 100x784 사이즈로 입력하는데 100은 배치, 78400(7만 얼마) 다 사용한걸 1 epoch 라고 함
# accuracy는 늘어나는게 맞고 loss는 줄어드는게 맞음 -> 제대로 학습시키고 있구나를 알 수 있음
# 두 값이 날뛸때는 뭔가 잘못되고 있으니 노드를 고치든, 손실함수를 고치든 해야됨
```
