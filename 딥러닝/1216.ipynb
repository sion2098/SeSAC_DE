1. Chaining  
-> 앞 단계의 출력 결과를 다음 단계의 입력(context)으로 넘기는 방식

````python
# OpenAI API를 동기 방식으로 쓰기
from openai import AsyncOpenAI, OpenAI # OpenAI: 동기 클라이언트 # AsyncOpenAI: 비동기 클라이언트

sync_client = OpenAI() # 동기 클라이언트 객체 생성 # 이 객체로 OpenAI API에 요청 보냄
````

````python
# 이 함수로 동작시킬거임
# 프롬프트 받아서 OpenAI API 호출하고 LLM의 답변 문자열을 돌려주는 함수
def llm_call(prompt:str, model:str = 'gpt-4o-mini') -> str:
  messages = [] # 여기에 OpenAI API가 받은 {role, content} 딕셔너리 넣어야 함
  messages.append({'role':'user','content':prompt})
  chat_completion = sync_client.chat.completions.create(model=model,messages=messages)
  return chat_completion.choices[0].message.content
````

````python
# 타입 힌트(Type Hint)를 쓰기 위한 준비 코드
# List 타입을 타입 힌트용으로 가져옴
# 타입 힌트 -> 이 변수/함수가 어떤 타입을 쓰는지 알려주는 설명
from typing import List
````

 <디버깅(확인용) 출력코드>  
 -> LLM이 각 단계에서 어떤 프롬프트를 실제로 받는지 눈으로 확인 가능  
 print(f'\n == 단계 {i} == \n')  
 print(f'프롬프트:\n{final_prompt}\n')

````python
# 여러 개의 프롬프트를 순서대로 실행하면서 이전 단계의 응답을 다음 단계의 문맥(context)으로 넘기는 체인 워크 플로우
def prompt_chain_workflow_2(initial_input:str, prompt_chain:List[str]) -> List[str]:
  response_chain=[] # 각 단계(프롬프트)의 LLM 응답을 순서대로 저장할 리스트

  # response : 현재 문맥(context) 역할을 하는 변수
  # 처음에는 사용자의 원래 입력을 그대로 문백으로 사용
  response=initial_input

  # prompt_chain에 들어있는 프롬프트(단계 지시문)를 1단계부터 순서대로 반복
  # i: 단계 번호(1,2,3,...)
  # prompt : 현재 단계에서 사용할 지시문(프롬프트 문자열)
  # enumerate(list) : 리스트의 값 + index를 동시에 꺼내주는 함수
  for i, prompt in enumerate(prompt_chain, 1): # prompt_chain에 들어있는 프롬프트를 하나씩 꺼내면서 단계 번호는 i, 프롬프트 내용은 prompt에 넣어서 출력, 번호는 1번부터 시작

    # final_prompt : LLM에 실제로 보낼 최종 프롬프트 문자열
    # prompt : 이번 단계에서 뭘 할지(지시문)
    # response : 이전 단계 결과 or 현재까지의 문맥
    # initial_input : 사용자가 맨 처음 준 원래 요청
    final_prompt = f'{prompt} \n\n 문맥(Context): \n{response}\n 사용자 입력 : {initial_input}'

    # llm_call(final_prompt) : OpenAI API를 호출해서 final_prompt에 대한 답변을 받음
    # 받은 답변을 response에 저장해서 다음 단계의 문맥(context)로 넘김
    response = llm_call(final_prompt)

    # 이번 단계의 결과를 response_chain에 저장
    response_chain.append(response)

   # 모든 단계가 끝나면, 각 단계의 응답 리스트 반환
  return response_chain
````

#### Chain workflow(체인 워크플로우)  
-> 앞 단계의 LLM 출력 결과를 다음 단계의 입력(context)으로 계속 넘기는 구조  

<체인, chain>
```
입력 → LLM → 결과1
결과1 → LLM → 결과2
결과2 → LLM → 결과3
```

2. Routing

````python
def run_router_workflow(user_prompt:str) :
    router_prompt = f"""
    사용자의 프롬프트/질문: {user_prompt}

    각 모델은 서로 다른 기능을 가지고 있습니다. 사용자의 질문에 가장 적합한 모델을 선택하세요:
    - gpt-4o: 일반적인 작업에 가장 적합한 모델 (기본값)
    - o1-mini: 코딩 및 복잡한 문제 해결에 적합한 모델
    - gpt-4o-mini: 간단한 사칙연산 등의 작업에 적합한 모델

    모델명만 단답형으로 응답하세요
    """
    # 각 모델이 어떤 상황에서 사용하면 되는지 미리 제시함

    selected_model = llm_call(router_prompt)
    print('llm이 선택한 모델 : ', selected_model)
    response = llm_call(user_prompt, model = selected_model)
    return response
````

````python
def run_router_workflow(user_prompt:str) :
    router_prompt = f"""
    사용자의 프롬프트/질문: {user_prompt}

    각 모델은 서로 다른 기능을 가지고 있습니다. 사용자의 질문에 가장 적합한 모델을 선택하세요:
    - gpt-3.5-turbo: 일반적인 작업에 가장 적합한 모델 (기본값)
    - gpt-4o: 코딩 및 복잡한 문제 해결에 적합한 모델
    - gpt-4o-mini: 간단한 사칙연산 등의 작업에 적합한 모델

    모델명만 단답형으로 응답하세요
    """

    selected_model = llm_call(router_prompt)
    print('llm이 선택한 모델 : ', selected_model)
    response = llm_call(user_prompt, model = selected_model)
    return response
````

````python
async def llm_call_async(prompt:str, model:str = 'gpt-4o-mini') -> str:
  messages = []
  messages.append({'role':'user','content':prompt})
  chat_completion = await client.chat.completions.create(
      model=model,
      messages=messages)
  print(model, '완료')
  return chat_completion.choices[0].message.content
````

````python
# 여러 LLM 요청을 동시에 실행하고 끝나는 순서대로 결과를 수집하는 비동기 병렬 처리
async def run_llm_parallel(prompt_details): # 비동기 함수 선언 # 비동기 함수 안에서 await 사용 가능
  tasks = [llm_call_async(prompt['user_prompt'], prompt['model']) for prompt in prompt_details] # 이벤트 루프에 등록되어 있는 코루틴
  responses = [] # LLM 응답 결과를 담을 빈 리스트(append로 추가)
  for task in asyncio.as_completed(tasks): # asyncio.as_completed(tasks) : 여러 코루틴 중 먼저 끝난 것부터 하나씩 반환(완료 순서대로 처리)
    result = await task # 하나의 코루틴(task)이 끝날 때까지 기다림 -> 끝나면 그 결과(LLM응답)을 result에 저장
    responses.append(result) # 완료된 LLM 응답을 리스트에 저장
  return responses # 모든 LLM 호출이 끝난 뒤 결과 리스트 반환
````

```
tasks = [llm_call_async(prompt['user_prompt'], prompt['model']) for prompt in prompt_details]
```
- prompt_details : 프롬프트 목록
- llm_call_async() : async 함수로 코루틴 객체가 만들어짐  
 -> tasks = [코루틴1, 코루틴2, 코루틴3, ...]  
 - prompt 는 딕셔너리(dict)
 - prompt['user_prompt'], prompt['model']
 -> 딕셔너리 안에 들어있는 값(value) 꺼내는 것


````python
async def main():
  question = ("아래 문장을 자연스러운 한국어로 번역해줘:\n"
                  "\"Do what you can, with what you have, where you are.\" — Theodore Roosevelt")

  parallel_prompt_details = [
          {"user_prompt": question, "model": "gpt-4o"},
          {"user_prompt": question, "model": "gpt-4o-mini"},
          {"user_prompt": question, "model": "o3"},
      ]

  responses = await run_llm_parallel(parallel_prompt_details)
  aggregator_prompt = ("다음은 여러 개의 AI 모델이 사용자 질문에 대해 생성한 응답입니다.\n"
                         "당신의 역할은 이 응답들을 모두 종합하여 최종 답변을 제공하는 것입니다.\n"
                         "일부 응답이 부정확하거나 편향될 수 있으므로, 신뢰성과 정확성을 갖춘 응답을 생성하는 것이 중요합니다.\n\n"
                         "사용자 질문:\n"
                         f"{question}\n\n"
                         "모델 응답들:")

  for i in range(len(parallel_prompt_details)):
    aggregator_prompt += f'\n{i+1}, 모델 응답 : {responses[i]}\n'

  print(aggregator_prompt)
  final_resp = await llm_call_async(aggregator_prompt, model='gpt-4o')
  print(final_resp)
````

````python
await main()
# 로컬에서 사용할 때
# asyncio.run(main()) 이렇게 써줘야 함
````
